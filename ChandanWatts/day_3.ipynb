## Part1

```python
count=0
obj_nos_true =[]
obj_nos_else =[]
count_e=0
for f in range(len(obj)):
    coln = obj[f]
    if coln['i_calib_psf_used'] == True or coln['i_calib_psf_reserved'] == True:
        obj_nos_true.append(f)
        count=count+1
    else:
        obj_nos_else.append(f)
        count_e = count_e+1
        
print(count, count_e)         
```

```python
flux_ratios_true=[]
x_par_true=[]
flux_ratios_else=[]
x_par_else=[]
flux_ratios_true_color=[]
x_par_true_color=[]
flux_ratios_else_color=[]
x_par_else_color=[]
for m in obj_nos_true:
    if obj['i_gaap0p7Flux'][m] > 0 and obj['i_gaap1p0Flux'][m] >0:
        flux_ratios_true.append(np.log10(obj['i_gaap0p7Flux'][m]/obj['i_gaap1p0Flux'][m]))
        x_par_true.append(np.log10(obj['i_gaap0p7Flux'][m]))
    if obj['r_gaap1p0Flux'][m] > 0 and obj['i_gaap1p0Flux'][m] >0 and obj['r_ap12Flux'][m] > 0 and obj['i_ap12Flux'][m] > 0:
        flux_ratios_true_color.append(np.log10(obj['r_gaap1p0Flux'][m]/obj['i_gaap1p0Flux'][m]))
        x_par_true_color.append(np.log10(obj['r_ap12Flux'][m]/obj['i_ap12Flux'][m]))
for m in obj_nos_else:
    if obj['i_gaap1p0Flux'][m] > 0 and obj['i_gaap0p7Flux'][m] >0:
        flux_ratios_else.append(np.log10(obj['i_gaap0p7Flux'][m]/obj['i_gaap1p0Flux'][m]))
        x_par_else.append(np.log10(obj['i_gaap0p7Flux'][m]))
    if obj['r_gaap1p0Flux'][m] > 0 and obj['i_gaap1p0Flux'][m] >0 and obj['r_ap12Flux'][m] > 0 and obj['i_ap12Flux'][m] > 0:
        flux_ratios_else_color.append(np.log10(obj['r_gaap1p0Flux'][m]/obj['i_gaap1p0Flux'][m]))
        x_par_else_color.append(np.log10(obj['r_ap12Flux'][m]/obj['i_ap12Flux'][m]))
```
