```python
import numpy as np
import pandas as pd
from astropy.io import fits
from astropy.table import Table
import matplotlib.pyplot as plt
import treecorr
```

```python
data = fits.open("/user/p5_lecturer/KiDS1000/KiDS_DR4.1_ugriZYJHKs_SOM_gold_WL_cat.fits")
    
```

```python
tab = Table.read(data[1])
```

```python
tab
```

```python
df = tab.to_pandas()
df
```

```python
bin1 = df[(df['Z_B']>0.1) & (df['Z_B']<=0.3)]
bin2 = df[(df['Z_B']>0.3) & (df['Z_B']<=0.5)]
bin3 = df[(df['Z_B']>0.5) & (df['Z_B']<=0.7)]
bin4 = df[(df['Z_B']>0.7) & (df['Z_B']<=0.9)]
bin5 = df[(df['Z_B']>0.9) & (df['Z_B']<=1.2)]
```

```python
bins = [bin1, bin2, bin3, bin4, bin5]


for i,b1 in enumerate(bins):

    for j,b2 in enumerate(bins):

        cat1 = treecorr.Catalog(ra = b1['RAJ2000'], dec = b1['DECJ2000'],ra_units='degrees', dec_units='degrees',
                           g1= b1['e1'], g2=b1['e2'], w = b1['weight'] )
    
        cat2 = treecorr.Catalog(ra = b2['RAJ2000'], dec = b2['DECJ2000'],ra_units='degrees', dec_units='degrees',
                           g1= b2['e1'], g2=b2['e2'], w = b2['weight'] )
        
    
        gg = treecorr.GGCorrelation(min_sep=0.5, max_sep=300., nbins=9,
                                sep_units='arcmin')
        gg.process(cat1, cat2)
        xip = gg.xip  # The xi_plus correlation function
        xim = gg.xim  # The xi_minus correlation function
        filename = f"corr_bin{i}_bin{j}.txt"
        gg.write(filename)
                           
```

```python
nbins = 5

fig, axes = plt.subplots(
    nbins, nbins,
    figsize=(15, 15),
    sharex=False,
    sharey=False
)

for i in range(nbins):
    for j in range(nbins):

        ax = axes[i, j]

        # Keep only the lower triangle
        if j > i:
            ax.axis('off')
            continue

        data = np.loadtxt(f"corr_bin{j}_bin{i}.txt")

        theta = data[:,1]
        xip   = data[:,3]
        xim   = data[:,4]
        sig_ip = data[:,7]
        sig_im = data[:,8]

        yip = theta * xip * 10**4
        yim = theta * xim * 10**4
        yip_err = theta * sig_ip * 10**4
        yim_err = theta * sig_im * 10**4


        # xi+
        ax.errorbar(theta, y=yip, yerr=yip_err, fmt='o', color='black', markersize=4, capsize=3, label=r'$\xi_+$')
     
        # xi-
        ax.errorbar(theta, y=yim, yerr=yim_err, fmt='o', color='red', markersize=4, capsize=3, label=r'$\xi_-$')
     

        # xi+
        #ax.scatter(theta, xip,s=18,marker='o',label=r'$\xi_+$')

        # xi-
        #ax.scatter(theta, np.abs(xim),s=18,marker='s,label=r'$\xi_-$')

        ax.set_ylim(-2.5, 5.5)
        ax.set_xscale('log')

        ax.set_title(f'({j+1},{i+1})', fontsize=10)

        ax.axhline(0, color='blue', linestyle='--', linewidth=1, alpha=0.7)

        ax.tick_params(labelsize=8)

        #ax.grid()

        # Label only left column
        if j == 0:
            ax.set_ylabel(r'$\Theta\xi_\pm$', fontsize=10)

        # Label only bottom row
        if i == nbins-1:
            ax.set_xlabel(r'$\theta$ (arcmin)', fontsize=10)

# One legend for the whole figure
handles, labels = axes[0,0].get_legend_handles_labels()
fig.legend(handles, labels,
           loc='upper right',
           fontsize=12)

plt.tight_layout()
plt.show()
```
