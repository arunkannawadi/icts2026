```python
rap_arr = [3,6,9,12,17,25,35,50,70]
mask_all_ap_psf_used = np.ones(len(obj), dtype=bool)

for r_ap in rap_arr:
    mask_all_ap_psf_used &=  (obj[f'i_ap{r_ap:02d}Flux_flag'] == False) &  (obj['i_calib_psf_used'] == True)

objgood_psf_used = obj[mask_all_ap_psf_used]
i=20

flux_good_psf_used = [objgood_psf_used[f'i_ap{r_ap:02d}Flux'][i] for r_ap in rap_arr]
Err_flux_good_psf_used = [objgood_psf_used[f'i_ap{r_ap:02d}FluxErr'][i] for r_ap in rap_arr]
print(len(flux_good_psf_used), len(Err_flux_good_psf_used) )
#print(flux_good_psf_used)
SN_ratio = np.array(flux_good_psf_used)/np.array(Err_flux_good_psf_used)
print(SN_ratio)
psf_flux = objgood_psf_used['i_psfFlux'][i]
psf_flux_error = objgood_psf_used['i_psfFluxErr'][i]
cmodel_flux = objgood_psf_used['i_cModelFlux'][i]
cmodel_flux_error = objgood_psf_used['i_cModelFluxErr'][i]
sersic_flux = objgood_psf_used['i_sersicFlux'][i]
sersic_flux_error = objgood_psf_used['i_sersicFluxErr'][i]
SN_ratio_psf = np.array(psf_flux)/np.array(psf_flux_error)
SN_ratio_cmodel = np.array(cmodel_flux)/np.array(cmodel_flux_error)
SN_ratio_sersic = np.array(sersic_flux)/np.array(sersic_flux_error)

print(SN_ratio_sersic)

plt.plot(rap_arr, SN_ratio, 'o-')
plt.axhline(SN_ratio_psf, color='k', ls='--',label='PSF')
plt.axhline(SN_ratio_cmodel, color='red', ls='--',label='cModel')
plt.axhline(SN_ratio_sersic, color='green', ls='--',label='Sersic')

plt.xlabel("Aperture radius (pixels)")
plt.ylabel("S/N ")
#plt.yscale('log')
#plt.grid(True)
plt.legend(fontsize=20)
plt.savefig('patch_69_tract_0_StoNratio_aperture.png', bbox_inches="tight", dpi=300)
plt.show()
```
